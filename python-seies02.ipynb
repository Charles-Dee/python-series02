```python
import matplotlib.pyplot as plt
import numpy as np

# Create time array (1000 values from 0 to 1)
t = np.linspace(0, 1, 1000)

# Create signal array (sine wave scaled for one complete cycle over t)
s = np.sin(2 * np.pi * t)

# Plot s as a function of t
plt.figure(figsize=(8, 4))
plt.plot(t, s)

# Add title, labels, and grid
plt.title("Sine Wave Signal Over Time")
plt.xlabel("Time (t)")
plt.ylabel("Amplitude (s)")
plt.grid(True)

# Save plot as a PNG file
plt.savefig("sine_wave_plot.png", dpi=300, bbox_inches="tight")

# Display the plot in the notebook
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Create time array (1000 values from 0 to 1)
t = np.linspace(0, 1, 1000)

# Create signal arrays for sine and cosine
s1 = np.sin(2 * np.pi * t)
s2 = np.cos(2 * np.pi * t)

# Initialize figure and primary y-axis
fig, ax1 = plt.subplots(figsize=(9, 5))

# Plot sine on the primary y-axis
line1 = ax1.plot(t, s1, color="tab:blue", label="Sine Signal (s1)")
ax1.set_xlabel("Time (t)")
ax1.set_ylabel("Sine Amplitude", color="tab:blue")
ax1.tick_params(axis="y", labelcolor="tab:blue")
ax1.grid(True, linestyle=":", alpha=0.6)

# Create a secondary y-axis sharing the same x-axis
ax2 = ax1.twinx()

# Plot cosine on the secondary y-axis
line2 = ax2.plot(
    t, s2, color="tab:orange", linestyle="--", label="Cosine Signal (s2)"
)
ax2.set_ylabel("Cosine Amplitude", color="tab:orange")
ax2.tick_params(axis="y", labelcolor="tab:orange")

# Combine legend items from both axes into a single box
lines = line1 + line2
labels = [line.get_label() for line in lines]
ax1.legend(lines, labels, loc="upper right")

# Set figure title
plt.title("Sine and Cosine Signals Over Time")

# Save the plot as a PDF file
plt.savefig("signals_plot.pdf", format="pdf", bbox_inches="tight")

# Display the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Create time array (1000 values from 0 to 1)
t = np.linspace(0, 1, 1000)

# Create primary signal (1 Hz sine wave)
s = np.sin(2 * np.pi * 1 * t)


# --- 1. Signal Plot with Secondary Y-Axis ---
fig, ax1 = plt.subplots(figsize=(8, 4))

# Primary Y-axis (left)
line1 = ax1.plot(t, s, color="tab:blue", label="sine")
ax1.set_xlabel("Time (s)")
ax1.set_ylabel("Sine Amplitude", color="tab:blue")
ax1.grid(True)

# Secondary Y-axis (right)
ax2 = ax1.twinx()
s2 = np.sin(2 * np.pi * 2 * t)  # 2 Hz sine wave
line2 = ax2.plot(t, s2, color="tab:orange", linestyle="--", label="Cosine")
ax2.set_ylabel("Cosine Amplitude", color="tab:orange")

# Combined legend for twin axes
lines = line1 + line2
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc="lower left")

plt.title("Sine Cosine Comparison")

# Save dual plot as PDF
plt.savefig("sine_and_cosine_wave.pdf", bbox_inches="tight")
plt.show()
```
