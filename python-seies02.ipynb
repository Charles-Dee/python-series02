- The goal of this assignment is to start coding something useful, from theory to code.

- The idea is to create a notebook to plot sinusoidal signals, in three different versions: **draft version**, a **better version** and a **final version**.


```python
import matplotlib.pyplot as plt
import numpy as np

# Create time array (1000 values from 0 to 1)
t = np.linspace(0, 1, 1000)

# Create signal array (sine wave scaled for one complete cycle over t)
s = np.sin(2 * np.pi * t)

# Plot s as a function of t
plt.figure(figsize=(8, 4))
plt.plot(t, s)

# Add title, labels, and grid
plt.title("Sine Wave Signal Over Time")
plt.xlabel("Time (t)")
plt.ylabel("Amplitude (s)")
plt.grid(True)

# Save plot as a PNG file
plt.savefig("sine_wave_plot.png", dpi=300, bbox_inches="tight")

# Display the plot in the notebook
plt.show()
```
